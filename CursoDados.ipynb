```python
turma_a = [1.6, 1.77, 1.82, 1.75, 2]
turma_b = [1.55, 1.60, 1.65, 2, 1.7]

#histograma - gráfico de distribuição
fig
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(turma_a, bins=5, kde=True, color='skyblue', label='Turma A')
plt.title('Distribuição de Alturas - Turma A')
plt.xlabel('Altura')
plt.ylabel('Frequência')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(turma_b, bins=5, kde=True, color='skyblue', label='Turma B')
plt.title('Distribuição de Alturas - Turma B')
plt.xlabel('Altura')
plt.ylabel('Frequência')
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Combine the data into a single DataFrame for easier plotting
data = {
    'Altura': turma_a + turma_b,
    'Turma': ['A'] * len(turma_a) + ['B'] * len(turma_b)
}
df_boxplot = pd.DataFrame(data)

plt.figure(figsize=(8, 6))
sns.boxplot(x='Turma', y='Altura', data=df_boxplot, palette={'A': 'skyblue', 'B': 'skyblue'})
plt.title('Comparação de Alturas entre Turma A e Turma B (Boxplot)')
plt.xlabel('Turma')
plt.ylabel('Altura (m)')
plt.grid(False)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='Turma', y='Altura', data=df_boxplot, showfliers=False, hue='Turma', palette={'A': 'skyblue', 'B': 'skyblue'}, legend=False)
plt.title('Comparação de Alturas entre Turma A e Turma B (Boxplot sem Outliers)')
plt.xlabel('Turma')
plt.ylabel('Altura (m)')
plt.grid(False)
plt.show()
```

```python
print(f'A média da turma A é: {np.mean(turma_a)} com desvio padrão de {np.std(turma_a)}')
print(f'A média da turma B é: {np.mean(turma_b)} com desvio padrão de {np.std(turma_b)}')
```

```python
print(f'A mediana da turma A é: {np.median(turma_a)} com desvio padrão de {np.std(turma_a)}')
print(f'A mediana da turma B é: {np.median(turma_b)} com desvio padrão de {np.std(turma_b)}')
```
